```python
from pandapower.create import (
    create_empty_network,
    create_bus,
    create_gen,
    create_load,
    create_line_from_parameters,
    create_transformer_from_parameters,
    create_ext_grid,
    create_shunt
)

from pandapower.run import runpp

import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
#read xlsx file
import pandas as pd

xls = pd.ExcelFile('ieee39bus.xlsx')

xls.sheet_names

sheet = {}

for sheet_name in xls.sheet_names:
    sheet[sheet_name] = xls.parse(sheet_name)
```

```python
net = create_empty_network()
```

```python
#read bus data
buses = {}

for i in range(len(sheet['bus'])):
    buses[i] = create_bus(net, vn_kv=sheet['bus']['vn_kv'][i],
                          name=sheet['bus']['name'][i],
                          type=sheet['bus']['type'][i],
                          in_service=sheet['bus']['in_service'][i],
                          max_vm_pu=sheet['bus']['max_vm_pu'][i],
                          min_vm_pu=sheet['bus']['min_vm_pu'][i])
net.bus
```

```python
#read generator data
gens = {}

for i in range(len(sheet['gen'])):
    gens[i] = create_gen(net, name=sheet['gen']['name'][i],
                         bus=sheet['gen']['bus'][i],
                         p_mw=sheet['gen']['p_mw'][i],
                         vm_pu=sheet['gen']['vm_pu'][i],
                         sn_mva=sheet['gen']['sn_mva'][i],
                         min_q_mvar=sheet['gen']['min_q_mvar'][i],
                         max_q_mvar=sheet['gen']['max_q_mvar'][i],
                         controllable=sheet['gen']['controllable'][i],
                         in_service=sheet['gen']['in_service'][i],
                         min_p_mw=sheet['gen']['min_p_mw'][i],
                         max_p_mw=sheet['gen']['max_p_mw'][i],
                        )
net.gen
```

```python
#read transformer data
trafos={}

for i in range(len(sheet['trafo'])):
    trafos[i] = create_transformer_from_parameters(net, name=sheet['trafo']['name'][i],
                             hv_bus=sheet['trafo']['hv_bus'][i],
                             lv_bus=sheet['trafo']['lv_bus'][i],
                             sn_mva=sheet['trafo']['sn_mva'][i],
                             vn_hv_kv=sheet['trafo']['vn_hv_kv'][i],
                             vn_lv_kv=sheet['trafo']['vn_lv_kv'][i],
                             vkr_percent=sheet['trafo']['vkr_percent'][i],
                             vk_percent=sheet['trafo']['vk_percent'][i],
                             pfe_kw=sheet['trafo']['pfe_kw'][i],
                             i0_percent=sheet['trafo']['i0_percent'][i],
                             vector_group=sheet['trafo']['vector_group'][i],
                             vk0_percent=sheet['trafo']['vk0_percent'][i],
                             in_service=sheet['trafo']['in_service'][i],
                             shift_degree=sheet['trafo']['shift_degree'][i],
                             tap_side=sheet['trafo']['tap_side'][i],
                             tap_neutral=sheet['trafo']['tap_neutral'][i],
                             tap_max=sheet['trafo']['tap_max'][i],
                             tap_min=sheet['trafo']['tap_min'][i],
                             tap_step_percent=sheet['trafo']['tap_step_percent'][i],
                             max_loading_percent=sheet['trafo']['max_loading_percent'][i]
                            )
net.trafo
```

```python
#read line data
lines={}

for i in range(len(sheet['line'])):
    lines[i]=create_line_from_parameters(net, name=sheet['line']['name'][i],
                                         from_bus=sheet['line']['from_bus'][i],
                                         to_bus=sheet['line']['to_bus'][i],
                                         length_km=sheet['line']['length_km'][i],
                                         r_ohm_per_km=sheet['line']['r_ohm_per_km'][i],
                                         x_ohm_per_km=sheet['line']['x_ohm_per_km'][i],
                                         c_nf_per_km=sheet['line']['c_nf_per_km'][i],
                                         max_i_ka=sheet['line']['max_i_ka'][i],
                                         in_service=sheet['line']['in_service'][i], 
                                         type=sheet['line']['type'][i],
                                         max_loading_percent=sheet['line']['max_loading_percent'][i]
                                        )
net.line
```

```python
#read load data
loads={}

for i in range(len(sheet['load'])):
    loads[i]=create_load(net, name=sheet['load']['name'][i],
                         bus=sheet['load']['bus'][i],
                         p_mw=sheet['load']['p_mw'][i],
                         q_mvar=sheet['load']['q_mvar'][i],
                         in_service=sheet['load']['in_service'][i],
                         type=sheet['load']['type'][i],
                         controllable=sheet['load']['controllable'][i]
                        )
net.load
```

```python
#create external grid
grids={}

for i in range(len(sheet['grid'])):
    grids[i]=create_ext_grid(net, name=sheet['grid']['name'][i],
                             bus=sheet['grid']['bus'][i],
                             vm_pu=sheet['grid']['vm_pu'][i],
                             va_degree=sheet['grid']['va_degree'][i],
                             max_p_mw=sheet['grid']['max_p_mw'][i],
                             min_p_mw=sheet['grid']['min_p_mw'][i],
                             max_q_mvar=sheet['grid']['max_q_mvar'][i],
                             min_q_mvar=sheet['grid']['min_q_mvar'][i],
                             in_service=sheet['grid']['in_service'][i],
                             controllable=sheet['grid']['controllable'][i]
                            )
net.ext_grid
```

```python
#run load flow
runpp(net)
```
