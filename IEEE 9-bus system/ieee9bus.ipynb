```python
from pandapower.create import (
    create_empty_network,
    create_bus,
    create_gen,
    create_load,
    create_line_from_parameters,
    create_transformer_from_parameters,
    create_ext_grid
)

from pandapower.run import runpp

import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
#create empty network
net = create_empty_network()
```

```python
#create buses
buses = {}

buses[1] = create_bus(net, vn_kv=16.5, name="Bus 1", type="b", in_service=True, max_vm_pu=1.1, min_vm_pu=0.90, zone="", geo="")
buses[2] = create_bus(net, vn_kv=18, name="Bus 2", type="b", in_service=True, max_vm_pu=1.1, min_vm_pu=0.90, zone="", geo="")
buses[3] = create_bus(net, vn_kv=13.8, name="Bus 3", type="b", in_service=True, max_vm_pu=1.1, min_vm_pu=0.90, zone="", geo="")

#create buses automatically
for i in range(4,10):
    buses[i] = create_bus(net, vn_kv=230, name=f"Bus {i}", index=i, type="b", in_service=True, max_vm_pu=1.1, min_vm_pu=0.90, zone="", geo="")

#data structure of the bus
net.bus
```

```python
#create external grid
grids = {}

grids[1] = create_ext_grid(net, name="grid", bus=buses[1], vm_pu=1.04, va_degree=0, max_p_mw=247.5, min_p_mw=0, max_q_mvar=247.5, min_q_mvar=-247.5, in_service=True, slack_weight=1/3, controllable=True)
#grids[2] = create_ext_grid(net, bus=buses[2], vm_pu=1.025, va_degree=9.3057, max_p_mw=163.2, min_p_mw=0, max_q_mvar=192, min_q_mvar=-192, in_service=True, slack_weight=1/3, controllable=True)
#grids[3] = create_ext_grid(net, bus=buses[3], vm_pu=1.025, va_degree=5.1420, max_p_mw=108.8, min_p_mw=0, max_q_mvar=128, min_q_mvar=-128, in_service=True, slack_weight=1/3, controllable=True)

#data structure of external grid
net.ext_grid
```

```python
#create transformer
trafos = {}
trafos[1] = create_transformer_from_parameters(net, hv_bus=buses[4], lv_bus=buses[1], sn_mva=250, vn_hv_kv=230, vn_lv_kv=16.5, vkr_percent=0.144, vk_percent=14.4, pfe_kw=0, i0_percent=0, vector_group='YNd', vk0_percent=3, in_service=True, name="Trafo 1", shift_degree=150, tap_side="hv", tap_neutral=0, tap_max=2, tap_min=-2, tap_step_percent=2.5, max_loading_percent=100, tap_step_degree=0.0, tap_changer_type="", id_characteristic_table=0.0, std_type="")
trafos[2] = create_transformer_from_parameters(net, hv_bus=buses[7], lv_bus=buses[2], sn_mva=200, vn_hv_kv=230, vn_lv_kv=18.0, vkr_percent=0.125, vk_percent=12.5, pfe_kw=0, i0_percent=0,vector_group='YNd', vk0_percent=3, in_service=True, name="Trafo 2", shift_degree=150, tap_side="hv", tap_neutral=0, tap_max=2, tap_min=-2, tap_step_percent=2.5, max_loading_percent=100, tap_step_degree=0.0, tap_changer_type="", id_characteristic_table=0.0, std_type="")
trafos[3] = create_transformer_from_parameters(net, hv_bus=buses[9], lv_bus=buses[3], sn_mva=150, vn_hv_kv=230, vn_lv_kv=13.8, vkr_percent=0.0879, vk_percent=8.79, pfe_kw=0, i0_percent=0,vector_group='YNd', vk0_percent=3, in_service=True, name="Trafo 3", shift_degree=150, tap_side="hv", tap_neutral=0, tap_max=2, tap_min=-2, tap_step_percent=2.5, max_loading_percent=100, tap_step_degree=0.0, tap_changer_type="", id_characteristic_table=0.0, std_type="")

#data structure of trafo
net.trafo
```
